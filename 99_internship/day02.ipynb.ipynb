```python
# use the pydantic to enforce data types
# no need of init(constructor)
# no need of str method also
from pydantic import BaseModel
class Student(BaseModel):
    name: str
    age: int
    marks : float
    
```

```python
st = Student(name='vishal', age=20, marks=90)
```

```python
print(st)
```

```python
# normal class
class Student1:
    def __init__(self, name : str, age : int, marks : float):
        self.name = name
        self.age = age
        self.marks = marks
    def __str__(self):
        return f"Student1(name={self.name}, age={self.age}, marks={self.marks})"
```

```python
st1 = Student1("vishal", 20, 90)
```

```python
print(st1)
```

```python
# by default it will be sorted based first value which is strings , alphabatically
sorted_vals = sorted([("name1", 90), ("name2", 100), ("name3", 70)])
print(sorted_vals)
```

```python
# to change it

def my_sortingbase(data):
    return data[1]

sorted_vals = sorted([("name1", 90), ("name2", 100), ("name3", 70)] , key= my_sortingbase)
print(sorted_vals)
```
