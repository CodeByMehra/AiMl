# Type Hinting and Annotations
#### 1. Basic function without type hints:

```python
def add(a,b):
    return a+b

print(add(2,4))
print(add("Heyy ","Coder")) # Works but types are unclear
```

```python
print(add(20,"coding")) # will give error
```

### 2 : Type Hints added

```python
def add(a: int,b: int)->int:  # type of a,b and -> return type
    return a+b
```

### 3. Variable Annotation

```python
name: str = "Vishal"
age: int = 22
```

### 4. Functions with multiple Types

```python
from typing import Union

def process_data(data: Union[int, str, float]) -> str:
    return f"processed {data}"

print(process_data(123))
print(process_data(1.23))
print(process_data(" hey brotha"))
```
