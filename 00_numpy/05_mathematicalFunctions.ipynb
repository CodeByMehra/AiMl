```python
import numpy as np

arr = np.array([1,2,3,4,5])
```

```python
# Aggregation
print("Sum:", np.sum(arr))
print("Product:", np.prod(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Std:", np.std(arr))
print("Variance:", np.var(arr))
print("Argmin:", np.argmin(arr))
print("Argmax:", np.argmax(arr))
```

```python
# Power
print("Square:", np.square(arr))
print("Square Root:", np.sqrt(arr))
print("Power:", np.power(arr, 2))
```

```python
# Log & Exponential
print("Log:", np.log(arr))
print("Log10:", np.log10(arr))
print("Log2:", np.log2(arr))
print("Exp:", np.exp(arr))
```

```python
# Rounding
print("Round:", np.round([4.3, 4.7]))
print("Floor:", np.floor([4.3, 4.7]))
print("Ceil:", np.ceil([4.3, 4.7]))
print("Trunc:", np.trunc([4.3, 4.7]))
```

```python
# Extras
arr2 = np.array([1, 2, 2, 3, 3, 4])
print("Unique:", np.unique(arr2))
print("Sort:", np.sort([5, 2, 8, 1]))
print("Absolute:", np.abs([-10, -20, 30]))
```
