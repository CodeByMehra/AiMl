```python
import numpy as np
import time
```

```python
#Execution Performance:

size= 1_000_000

# Python list:
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Python list time = {end-start} seconds")

#numpy arrays:

np_arr = np.array(py_list)
start = time.time()
#vectorization: 
sq_array = np_arr ** 2
end = time.time()
print(f"Python np_arr time = {end-start} seconds")
```
