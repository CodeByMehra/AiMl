## Perceptron:
 A perceptron is simple single layer neural network

- input (x1, x2 ....... xn)
- weight : A value added to each input feature is know as weights. generally these are initialized by 0
- Bias : it is a constant term that we add in the perceptron.


### Operations performed on perceptron:
## 1. weighted sum :
\[
z = \sum_{i=1}^{n} w_i x_i + b
\]

### Explanation:
- \(y\): Output
- \(w_i\): Weight of the \(i^{th}\) input
- \(x_i\): Input values
- \(n\): Number of inputs
- \(b\): Bias (constant added to shift the output)

### Expanded Form:
\[
z = w_1 x_1 + w_2 x_2 + w_3 x_3 + \cdots + w_n x_n + b
\]


## 2. Activation function:


- Activation function is applied on the z value produced by weighted sum
- eg: sigmoid, tanh, ReLu, softmax, Linear
