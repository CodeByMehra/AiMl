# DBSCAN

DBSCAN stands for Density Based Clustering of Applications with Noise

it is also used for anomalies detection/ outlier detection

- it is also good in separating non linear data

- we dont have to choose the value of k

- it divides in dataset in three categoris:

- 1. Core Point: A data point that has at least MinPts points within its ε (epsilon) neighborhood, meaning it lies in a dense region.

  2. Border Point: A data point that has fewer than MinPts neighbors within ε, but lies within the ε-neighborhood of a core point.

  3. Noise (Outlier): A data point that is neither a core point nor within the ε-neighborhood of any core point, so it does not belong to any cluster.

- to do this categorical classification, 2 imp hyperparameters are used:

- 1. epsilon value -> radius
  2. minimum-samples -> defines the density of points to make there a neighbourhood
