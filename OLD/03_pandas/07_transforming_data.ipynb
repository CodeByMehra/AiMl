```python
import pandas as pd
df = pd.read_csv("raw_data.csv")
#How do we transform our data to do feature engineering?
#Feature engineering: it is basically tranforming data to create new columns that improve performance of our models
#there are following main functions for this:

#1. apply() 
#2. map()
#3. assign()
#4. replace (old, new)
```

```python
df2 = df.copy()
df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 50000 else "10%")
df2
```

```python
#map():
#suppose we want to convert all male values by M, female by F and unknown by U

gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U"}
df2["gender"] = df2["gender"].map(gender_map)
df2
```

```python
#assign:
#increasing income byy 10%:

df2 = df2.assign(new_income = df2["income"] * 1.1)
df2
```

```python
#replace:

df2["country"] = df2["country"].replace("USA", "US")
df2
```
