```python
import matplotlib.pyplot as plt
```

```python
# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

plt.bar(oscar_years, oscar_revenue, color = "green")
plt.title("revenue of oscar movies")
plt.xlabel("Year")
plt.ylabel("revenue")

#showing revene at each bar
#plt.text(x-axis pos, y axis positon, text, alignment as  ha)
plt.text(oscar_years[0], oscar_revenue[0], "oscar_revenue", ha = "center")

plt.show()
```
