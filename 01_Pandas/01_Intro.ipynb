```python
import pandas as pd  # to import
```

there are two important data types of pandas:
1. Series -  it is like 1D Array
2. Dataframe - Similiar to 2D array
   

```python
# Series:

s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(s[0])
```

```python
s = pd.Series([18, 22, 23, 24, 35] ,  index=["Adam", "Vishal", "Bob", "Steive", "Alina"])  #providing custome levels
print(s)
print(s[0])  #-index based access
print(s["Vishal"])  # label based access
```

# DataFrame

```python
#creation using dictionary:

info = {
    "Name" : ["Adam", "Vishal", "Bob", "Steive", "Alina"],
    "Age" : [23, 22, 19, 21, 35],
    "GPA" : [9.6, 7.2, 8, 5.5 ,6]
}

df = pd.DataFrame(info)

print(df)
```

```python
#Cretion usning lists
df = pd.DataFrame([["adam", 21], ["bob", 21]], columns = [" Name", "Age"])
print(df)
```
