```python
import pandas as pd
```

```python
df = pd.read_csv("IRIS.csv")
df.head(10) # First 10 rows
```

```python
df.describe()
```

```python
result = df.groupby('species').agg({
    'sepal_length': 'mean',   # Average sepal_length
    'petal_width': 'max',     # Maximum petal_width
    'sepal_width': 'std'      # Standard deviation of sepal_width
})

print(result)
```
