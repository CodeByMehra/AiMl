```python
import pandas as pd

# ==========================================
# 1. SETUP: CREATING THE DATAFRAME
# ==========================================

# We start with "Wide Format" data.
# This is readable for humans but often bad for machines/plotting libraries.
df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22],
    "tax": [10, 20, 30, 40]
})

print("--- Original Wide DataFrame ---")
print(df)
```

```python
# ==========================================
# 2. MELT: WIDE TO LONG FORMAT
# ==========================================

# THEORY:
# Melt is used to "unpivot" data. It transforms columns into rows.
# Why? Many visualization libraries (like Seaborn) and ML models prefer "Long" format
# where each row is a single observation of a specific variable.

melted_df = df.melt(
    id_vars=["country", "year"],      # KEEP: These columns stay vertical (Identifiers)
    value_vars=["sales", "profit", "tax"], # UNPIVOT: These columns are stacked into rows
    var_name="metrics",               # NEW NAME: What to call the column containing 'sales', 'profit', etc.
    value_name="value"                # NEW VALUE: What to call the column containing the numbers
)

print("\n--- Melted (Long) DataFrame ---")
print(melted_df.head(6)) 
# Notice how 'sales', 'profit', and 'tax' are no longer columns, 
# but values inside the 'metrics' column.


```

```python
# ==========================================
# 3. PIVOT: LONG TO WIDE FORMAT (The Inverse)
# ==========================================

# THEORY:
# Pivot is the opposite of Melt. It takes a "Long" dataset and widens it.
# It spreads unique values from a column into multiple new column headers.
# This is usually done to make the final report "human-readable" again.

pivoted_df = melted_df.pivot(
    index=["country", "year"],  # KEEP: What should stay as the row identifiers?
    columns="metrics",          # HEADERS: Which column values become new column names?
    values="value"              # VALUES: What numbers fill the cells?
)

# Optional: Reset index to flatten the hierarchy and make it look like the original
pivoted_df = pivoted_df.reset_index()

# Remove the confusing 'metrics' axis label if it appears
pivoted_df.columns.name = None 

print("\n--- Pivoted (Back to Wide) DataFrame ---")
print(pivoted_df)
```

```python
# ==========================================
# 4. OTHER TRANSFORMATION TOOLS (From Slide 2)
# ==========================================

# A. Renaming Columns
# format: { "old_name": "new_name" }
df_renamed = df.rename(columns={"sales": "total_sales", "profit": "net_profit"})

# B. Sorting Values
# Sort by country (A-Z) then by sales (High to Low)
df_sorted = df.sort_values(by=["country", "sales"], ascending=[True, False])

# C. Ranking
# Assigns a rank based on value. 
# method='dense' means ties get the same rank, and the next rank is consecutive (1, 2, 2, 3)
df["sales_rank"] = df["sales"].rank(ascending=False, method="dense")

print("\n--- Transformed DataFrame (Sorted & Ranked) ---")
print(df_sorted)
```
