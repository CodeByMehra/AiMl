```python
#Method 2 : Web Scraping (THEORY FILE)
# Web scraping is the automated process of extracting data from websites using software or scripts.

# -----------------------------
# HTML FOR DATA SCIENCE
# -----------------------------
# HTML (HyperText Markup Language) is used to create and structure web pages.
# Web pages are built using tags that define elements like:
# headings, paragraphs, links, images, tables, lists, and forms.
# Browsers interpret HTML to display content.

# -----------------------------
# BASIC HTML STRUCTURE
# -----------------------------
# <!DOCTYPE html>  -> Defines the document as HTML5
# <html>           -> Root element of the page
# <head>           -> Contains metadata, title, scripts, styles
# <body>           -> Visible content of the page
# <h1>             -> Heading (largest)
# <p>              -> Paragraph

# Example structure:
# <html>
#   <head><title>My First Page</title></head>
#   <body>
#       <h1>Welcome!</h1>
#       <p>This is a paragraph.</p>
#   </body>
# </html>

# -----------------------------
# IMPORTANT HTML TAGS (SCRAPING)
# -----------------------------
# <h1> to <h6>   -> Headings (titles, sections)
# <p>            -> Paragraph text
# <a>            -> Anchor tag (hyperlinks)
# <img>          -> Images (use src attribute)
# <table>        -> Tables (rows = <tr>, columns = <td>/<th>)
# <ul>, <ol>     -> Lists
# <li>           -> List items
# <div>, <span>  -> Containers (used with class/id for scraping)
# <form>         -> Forms (input, button, submit)

# HTML follows a hierarchical (tree) structure.
# Tags can contain other tags (parent-child relationship).

# -----------------------------
# HTML ATTRIBUTES
# -----------------------------
# Attributes provide extra information about elements.
# Common attributes:
# href   -> URL in <a> tag
# src    -> Image source in <img> tag
# id     -> Unique identifier
# class  -> Group of elements (very useful in scraping)
# type   -> Defines input type (text, submit, etc.)
# name   -> Name of form element
# value  -> Data associated with form element

# -----------------------------
# PYTHON LIBRARIES FOR SCRAPING
# -----------------------------
# requests       -> Downloads web pages (HTML)
# BeautifulSoup  -> Parses & extracts data from HTML (static pages)
# Selenium       -> Automates browser for dynamic (JS-based) websites

# Usually used together:
# requests + BeautifulSoup

# -----------------------------
# WORKING WITH REQUESTS
# -----------------------------
# requests allows sending HTTP requests (GET, POST, etc.)
# Most common for scraping: GET request

# Install:
# pip install requests

# Import:
# import requests

# -----------------------------
# BASIC GET REQUEST
# -----------------------------
# response = requests.get("https://example.com")
# response.status_code -> HTTP status code
# response.text        -> HTML content (string)
# response.content     -> Raw bytes
# response.headers     -> Metadata

# ALWAYS check status_code before using data.

# Common status codes:
# 200 -> OK
# 301/302 -> Redirect
# 400 -> Bad Request
# 401 -> Unauthorized
# 403 -> Forbidden
# 404 -> Not Found
# 500 -> Server Error

# -----------------------------
# SAVING SCRAPED DATA
# -----------------------------
# with open("data.html", "w") as f:
#     f.write(response.text)

# -----------------------------
# HEADERS IN WEB SCRAPING
# -----------------------------
# Some websites block bots.
# Use headers like User-Agent to mimic a real browser.

# Example:
# headers = {"User-Agent": "Mozilla/5.0"}
# requests.get(url, headers=headers)

# -----------------------------
# ERROR HANDLING (try-except)
# -----------------------------
# Use try-except to handle network errors safely.

# try:
#     r = requests.get(url, timeout=5)
#     r.raise_for_status()
# except requests.exceptions.RequestException as e:
#     print("Error:", e)

# -----------------------------
# USING DELAYS
# -----------------------------
# Use time.sleep(seconds) to avoid sending requests too fast
# This helps prevent getting blocked.

# -----------------------------
# BEAUTIFULSOUP (bs4)
# -----------------------------
# BeautifulSoup parses HTML/XML and creates a parse tree.
# Used to extract:
# text, attributes, links, tables, etc.

# Install:
# pip install beautifulsoup4

# Import:
# from bs4 import BeautifulSoup

# -----------------------------
# PARSE TREE STRUCTURE
# -----------------------------
# html
# ├── head
# │   └── title
# └── body
#     ├── h1
#     └── p

# =========================================================
# END OF NOTES
# =========================================================
```
