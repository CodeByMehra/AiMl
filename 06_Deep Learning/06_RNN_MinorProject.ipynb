# Pre Processing

### 1. Converting to Lowercase:

```python
df["review"]= df["review"].str.lower()  # The strings that are available in the DF will be converted to Lowercase
```

### 2. Removing the URLs

```python
# To perform this action, we use oprn source library of pythom named re (regular expression)

import re

def remove_urls(text):
    text = re.sub(r"http\S+", "" , text)  # (pattern, replacement, string)
    return text
             # \S -> Non whitespace characters after the text (capital S)
             # \s -> whitespace character

df["review"] = df["review"].apply(remove_urls)
```

### 3. Remove punctuations:

```python
# We will consider the values as punctuations that does not comes under the range A-Z a-z 0-9 and \s (space)
def remove_punctuations(text):
    text = re.sub(r"^[A-Za-z0-9\s]", "" , text) # ^ means include
    return text

df["review"] = df["review"].apply(remove_punctuations)         
```

### 4, Removing HTML tags

```python
# Some symbols in RegEx:
# . -> any character
# * -> 0 or more character

def remove_html(text):
    text = re.sub(r"<.*?>", "" , text)
    return text

df["review"] = df["review"].apply(remove_html)      
```

### 5. Remove Stopwords

```python
import nltk 
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
def remove_stopwords(text):
    tokens = word_tokenize(text)
    stop_words = stopwords.words("english")

    for word in tokens:
        if word in stop_words:
            text = text.replace(word, "")

    return text

df["review"] = df["review"].apply(remove_stopwords)  
```
