```python
import torch
import torch.nn as nn 
```

```python
# creating input:
torch.manual_seed(42)  # Just like random state
inputs = torch.tensor([1.0, 2.0, 3.0])  # float format is necessary
```

```python
neuron = nn.Linear(in_features=3, out_features=1) # Defined input and output features
```

```python
output = neuron(inputs)
```

```python
output
```

```python
# Veriying the outpu with formula:
# submission of input and weight upto n

# @ is used for dot product

neuron.weight @ inputs + neuron.bias
```

```python
# Both values are equal -> Correct
```
