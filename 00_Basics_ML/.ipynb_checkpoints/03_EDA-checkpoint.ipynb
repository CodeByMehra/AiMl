EDA = Exploratory Data Analysis

This is where you build intuition about the dataset.

You analyze:
- Feature distributions
- Relationship with target
- Correlations
- Hidden patterns

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create simple example dataset
data = {
    "Age": [22, 25, 30, 35, 40, 28, 32, 45],
    "Salary": [20000, 25000, 30000, 35000, 50000, 27000, 32000, 60000],
    "Gender": ["M", "F", "F", "M", "M", "F", "M", "F"],
    "Purchased": [0, 0, 1, 1, 1, 0, 1, 1]   # Target variable
}

df = pd.DataFrame(data)
df
```

Feature Distributions:


Understand how values are spread in each feature.

```python
#1. Histogram (Numerical Features)


plt.hist(df["Age"], bins=5)
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

Shows how Age values are distributed:
```

#2 Relationship with Target


Now we check how features relate to target (Purchased).


✅ Numerical Feature vs Target

```python

sns.boxplot(x="Purchased", y="Age", data=df)
plt.title("Age vs Purchased")
plt.show

#Compare Age distribution for buyers vs non-buyers.
```

```python
#Categorical Feature vs Target
sns.countplot(x="Gender", hue="Purchased", data=df)
plt.title("Gender vs Purchased")
plt.show()
```

3 Correlations

Correlation shows strength of relationship between numerical features.
Range:


-1 → strong negative

 0 → no relation

+1 → strong positive

```python
corr = df.corr(numeric_only=True)
corr
```

```python
#Heatmap visualization:

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

4 .Finding Patterns


Scatter plots help identify patterns between two variables.

```python
plt.scatter(df["Age"], df["Salary"])
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title(" Age vs Salary")
plt.show()
```
