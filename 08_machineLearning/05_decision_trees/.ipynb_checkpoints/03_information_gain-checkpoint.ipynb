# 📌 Information Gain and Gini Gain (Decision Trees)

## 1️⃣ Information Gain (IG)

Information Gain is used in Decision Trees to decide the best feature to split the data.
It is based on **Entropy**, which measures impurity (disorder) in the dataset.

### 🔹 Entropy Formula

\[
Entropy(S) = - \sum_{i=1}^{c} p_i \log_2(p_i)
\]

Where:
- \( p_i \) = proportion of class i
- c = number of classes

Entropy range:
- 0 → Pure node (all samples same class)
- 1 → Maximum impurity (binary case)

### 🔹 Information Gain Formula

\[
IG(S, A) = Entropy(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} Entropy(S_v)
\]

Where:
- S = dataset
- A = attribute
- \( S_v \) = subset after split on attribute A

👉 Higher Information Gain = Better split.

Used in: **ID3 Algorithm**

---

## 2️⃣ Gini Gain (Based on Gini Index)

Gini Index measures impurity similar to entropy but is computationally simpler.

### 🔹 Gini Index Formula

\[
Gini(S) = 1 - \sum_{i=1}^{c} p_i^2
\]

Range:
- 0 → Pure node
- 0.5 → Maximum impurity (binary case)

### 🔹 Gini Gain

Instead of Information Gain, CART algorithm uses minimum weighted Gini:

\[
Gini_{split} = \sum_{v \in Values(A)} \frac{|S_v|}{|S|} Gini(S_v)
\]

👉 Choose attribute with **lowest Gini split value**.

Used in: **CART Algorithm**

---

## 🔎 Difference Between IG and Gini

| Feature | Information Gain | Gini Gain |
|----------|------------------|------------|
| Based on | Entropy | Gini Index |
| Log calculation | Yes | No |
| Computational cost | Higher | Lower |
| Used in | ID3 | CART |

---

## ✅ Summary

- Both are impurity measures for Decision Trees.
- Information Gain → Maximize reduction in entropy.
- Gini Gain → Minimize Gini impurity.
- Both aim to create pure child nodes.
