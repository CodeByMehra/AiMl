# 🌳 Pruning in Decision Trees (DT)

## 📌 What is Pruning?

Pruning is the process of **reducing the size of a Decision Tree** by removing unnecessary branches to avoid **overfitting**.

👉 Overfitting: Model performs very well on training data but poorly on test data.

Pruning improves:
- Generalization
- Model simplicity
- Prediction accuracy on unseen data

---

## 🔹 Why Pruning is Needed?

- Deep trees memorize noise.
- Complex trees have high variance.
- Reduces computational complexity.

---

## 🔸 Types of Pruning

### 1️⃣ Pre-Pruning (Early Stopping)

Stops tree growth before it becomes too complex.

Conditions used:
- Maximum depth limit
- Minimum samples per split
- Minimum Information Gain threshold

✔ Faster  
✖ May underfit if stopped too early

---

### 2️⃣ Post-Pruning (Backward Pruning)

Tree is fully grown first, then unnecessary branches are removed.

Steps:
1. Build full tree.
2. Remove nodes that do not improve validation accuracy.

✔ More reliable  
✔ Reduces overfitting effectively

---

## 🔹 Cost Complexity Pruning (CART)

Used in CART algorithm.

\[
Cost(T) = Error(T) + \alpha \times |T|
\]

Where:
- \( Error(T) \) = Misclassification error
- \( |T| \) = Number of leaf nodes
- \( \alpha \) = Complexity parameter

👉 Larger α → Smaller tree  
👉 Choose α using cross-validation.

---

## 🔎 Advantages of Pruning

- Reduces overfitting
- Improves model generalization
- Makes tree interpretable
- Reduces model size

---

## ⚖️ Bias-Variance Impact

- Without pruning → Low bias, High variance
- With pruning → Slightly higher bias, Lower variance

---

## ✅ Summary

- Pruning removes unnecessary branches.
- Two types: Pre-pruning and Post-pruning.
- Cost-complexity pruning balances error and tree size.
- Helps achieve optimal bias-variance tradeoff.
