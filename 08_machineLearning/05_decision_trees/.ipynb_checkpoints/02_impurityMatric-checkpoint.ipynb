Impuurity Metrics : Entropy

Entropy measures the impurity, disorder or ncertainity in a dataset
High entropy refers to the high ncertainity in the dataset, so it is good for the data set to have low Entropy


# How Entropy is measured?

Entropy(S) = - Σ (p_i * log₂(p_i))

Where:
- S = Dataset
- p_i = Probability of class i
- log₂ = Log base 2
- i = Number of classes
