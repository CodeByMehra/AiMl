## 📌 What is Terminal?

A **Terminal** is a program that allows users to interact with the computer using **text-based commands** instead of a graphical interface.

Instead of clicking buttons, you type commands to perform tasks like:
- Creating files
- Navigating folders
- Running programs

👉 It is widely used by developers for speed and control.

## 📖 Important Terms

### 1. Command Line
An interface where users type textual commands to interact with the system.

### 2. Terminal
A type of command-line interface used to execute commands.

### 3. Console
A command-line interface used to work directly with the computer system.

### 4. Shell
A program that runs inside the terminal and interprets commands given by the user.

### 5. Bash (Bourne Again Shell)
A popular shell used in Linux and macOS systems.

### 6. C Shell (csh)
Another type of shell, often used as a default in some systems.
