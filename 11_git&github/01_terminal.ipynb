## 📌 What is Terminal?

A **Terminal** is a program that allows users to interact with the computer using **text-based commands** instead of a graphical interface.

Instead of clicking buttons, you type commands to perform tasks like:
- Creating files
- Navigating folders
- Running programs

👉 It is widely used by developers for speed and control.

## 📖 Important Terms

### 1. Command Line
An interface where users type textual commands to interact with the system.

### 2. Terminal
A type of command-line interface used to execute commands.

### 3. Console
A command-line interface used to work directly with the computer system.

### 4. Shell
A program that runs inside the terminal and interprets commands given by the user.

### 5. Bash (Bourne Again Shell)
A popular shell used in Linux and macOS systems.

### 6. C Shell (csh)
Another type of shell, often used as a default in some systems.

## ⚙️ Basic Commands

| Command | Description |
|--------|------------|
| ls     | Lists files and directories |
| pwd    | Shows current working directory | 
| clear  | Clears the terminal screen |

## 🧭 Navigation Commands

### cd (Change Directory)
Used to move between folders.

Example:
cd folder_name

### Paths

#### Absolute Path
Full path from root directory.
Example:
C:/Users/Name/Documents

#### Relative Path
Path relative to current directory.
Example:
../folder_name

## 📂 Directory & File Commands

### mkdir
Creates a new directory.

Example:
mkdir myfolder

### Flags
Options added to commands to modify behavior.

Example:
ls -l   (detailed list)

---

### touch
Creates a new empty file.

Example:
touch file.txt

---

### rm
Deletes a file.

Example:
rm file.txt

---

### rmdir
Deletes an empty directory.

Example:
rmdir folder_name

---

### rm -rf
Forcefully deletes files and directories.

⚠️ Warning: This is a dangerous command.
