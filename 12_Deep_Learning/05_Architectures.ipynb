### CNN Architecture:
#### 1. Convolutional layer :
- Min- Max scaling:
        - Min-Max scaling is a normalization technique that transforms data into a fixed range, usually [0,1]. It rescales features by subtracting the minimum value and dividing by the range (max − min). This helps algorithms converge faster and ensures all features contribute equally.
- Applying Filter(kernel) :
      - Applying a filter (kernel) involves sliding a small matrix over an image to perform operations like blurring, sharpening, or edge detection. Each pixel value is recalculated using weighted sums of neighboring pixels. It is widely used in image processing and convolutional neural networks (CNNs).

- Padding :
          - Padding is added in image to keep the image information conserved
