```python
import torch
import torch.nn as nn
```

```python
# input

torch.manual_seed(42) # just like random state
inputs = torch.tensor([1.0, 2.0, 3.0])   # creation of vecotor as input
```

```python
neuron = nn.Linear(in_features=3, out_features=1)
```

```python
output = neuron(inputs)
```

```python
output
```

```python
# Verify:

# using formula of weighted sum
neuron.weight @ inputs + neuron.bias
```
