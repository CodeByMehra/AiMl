## Optimizers:
 these are algos that update the weight and bias to  minimize the loss/cost function


 most famous is gradient decsent

 1. Batch GD: in this type, the whole training data will be feeded as a single batch
 2. stochastic GD: here 1 batch = 1 sample
 3. Mini batch GD: in this , a mini batch size is taken


## Modern versions of GD:

1. GD with momentum:
     here we try to add the momentum in normal Gd


   the velocity term is calculated in this

2. RMS Prop:
   Root mean square propagation


   it is basically an improvement of Adagrad

3. ADAM (Adaptive moment estimation):

   it is the combination of upper two types


   most commonly used
