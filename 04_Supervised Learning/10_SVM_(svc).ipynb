#### WE are going to perform multiclass classification on IRIS Dataset

```python
from sklearn.svm import SVC
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
```

```python
df = datasets.load_iris(as_frame=True).frame
```

```python
df.head()
```

```python
X = df.drop("target",  axis=1)
y = df["target"]
```

```python
# Scaling data is very important in SVM
# in classification - scale input features only
# in regression - scale input and output features
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y) 
# stratify used for imbalanced classes 
```
