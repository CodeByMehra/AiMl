## Methods to choose the K value:

# 1. Elbow Method:

The **Elbow Method** is a simple technique used to determine the optimal number of clusters (**K**) in the **K-Means clustering algorithm**.

### Basic Idea
When we increase the number of clusters (**K**), the **Within-Cluster Sum of Squares (WCSS)** or **inertia** decreases because points are closer to their cluster centers. However, after a certain point, the improvement becomes very small.

### Steps
1. Run K-Means for a range of K values (e.g., 1–10).
2. Calculate the **WCSS (sum of squared distances from each point to its centroid)** for each K.
3. Plot **K vs WCSS**.
4. Look for the **"elbow point"** where the decrease in WCSS starts to slow down.

### Why It's Called the Elbow
The graph typically looks like an arm, and the point where the curve bends sharply (like an elbow) suggests the **optimal number of clusters**.

### Intuition
- Before the elbow: Adding clusters significantly improves clustering.
- After the elbow: Adding more clusters gives only small improvements.

### Conclusion
The **K value at the elbow point** is chosen because it balances **model simplicity** and **clustering accuracy**.
