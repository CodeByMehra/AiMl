## methods to choose K:

1. Elbow method:

in this methis WCSS (within cluster sum of squares) is calculated for each k(1,2,3.....n)

WCSS is the sum of squared distance of each point from the nearest centroid



2. Silhoutte score:

   silhotte score value s(i) differ in the range -1 to 1

   higher the value -> better clustering

   in this method, the silhoutte score is calculated for each k and the k with the highest value is choosen
