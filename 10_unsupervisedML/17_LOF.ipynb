# LOF - Local Outlier Factor

the points that are near to a cluster are known as local outliers

LOF algo is used to detect the local outliers


### steps:

- choose k (number of nearest neighbors)

- find the k-nearest neighbors of each data point

- compute the reachability distance based on neighbors

- calculate the local reachability density (LRD) of each point

- compute the LOF score by comparing density with neighbors (higher LOF ⇒ more outlier)
