# Wholesale Customers Dataset – Clustering Project Guide

## Dataset Link
https://archive.ics.uci.edu/ml/datasets/wholesale+customers

---

## 1. Dataset Description

- Total Records: 440 customers  
- Type: Unsupervised (no labels)  
- Domain: Wholesale distribution  

Each row represents:
> Annual spending of a customer across different product categories

---

## 2. Features Overview

### Categorical Features
- **Channel**
  - 1 → Horeca (Hotel/Restaurant/Cafe)
  - 2 → Retail

- **Region**
  - 1 → Lisbon
  - 2 → Oporto
  - 3 → Other

---

### Numerical Features (Main for Clustering)

These represent **annual spending**:

- Fresh → Fruits, vegetables  
- Milk → Dairy products  
- Grocery → Daily essentials  
- Frozen → Frozen food  
- Detergents_Paper → Cleaning products  
- Delicassen → Luxury/special items  

---

## 3. Problem Understanding

Each customer has a **spending pattern**.

Goal:
> Identify groups of customers with similar purchasing behavior

---

## 4. Project Aim

### Main Objective
Segment customers based on their purchasing behavior using clustering.

---

### Expected Output

You should be able to form clusters like:

- High Grocery + Detergents → Retail stores  
- High Fresh products → Restaurants  
- Low overall spending → Small customers  
- Balanced buyers → Mixed customers  

---

## 5. ML Pipeline (Follow Strictly)

### Step 1: Data Preprocessing
- Check missing values  
- Handle categorical features (encode or drop if needed)  

---

### Step 2: Outlier Detection
- Use boxplot or IQR  
- Remove extreme values  

---

### Step 3: Feature Scaling
- Use StandardScaler or MinMaxScaler  
- Required because feature values vary significantly  

---

### Step 4: Correlation Analysis
- Identify relationships:
  - Grocery ↔ Detergents  
  - Milk ↔ Grocery  

---

### Step 5: Find Optimal K
- Elbow Method  
- Silhouette Score  

---

### Step 6: Apply Clustering
- Use K-Means  

---

### Step 7: Cluster Analysis (Most Important)

For each cluster:
- Calculate mean values  
- Interpret behavior  

---

## 6. Final Deliverables

### 1. Cluster Visualization
- Scatter plots / PCA
