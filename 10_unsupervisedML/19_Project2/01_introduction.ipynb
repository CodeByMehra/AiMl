# Online Retail II Dataset – Clustering Project Guide

## Dataset Link
https://www.kaggle.com/datasets/mashlyn/online-retail-ii-uci

---

## 1. Dataset Description

- Type: Transactional dataset (E-commerce)
- Nature: Unsupervised (no target label)
- Size: ~500,000+ records
- Domain: Online retail store (UK-based)

Each row represents:
> A single transaction (purchase of a product by a customer)

---

## 2. Features Overview

- **InvoiceNo** → Unique transaction ID  
- **StockCode** → Product ID  
- **Description** → Product name  
- **Quantity** → Number of items purchased  
- **InvoiceDate** → Date and time of purchase  
- **UnitPrice** → Price per item  
- **CustomerID** → Unique customer identifier  
- **Country** → Customer location  

---

## 3. Important Observations

- Some transactions have:
  - Missing CustomerID  
  - Negative Quantity (product returns)  
- Dataset is not directly usable → requires preprocessing  
- Data is at transaction level, not customer level  

---

## 4. Problem Understanding

We do NOT directly cluster this raw data.

Instead:
> We need to convert transaction data into customer behavior data

This is done using **RFM Analysis**.

---

## 5. Feature Engineering (Core Concept)

We create new features:

- **Recency (R)**  
  → How recently a customer made a purchase  

- **Frequency (F)**  
  → How often a customer makes purchases  

- **Monetary (M)**  
  → Total amount spent by a customer  

---

## 6. Project Aim

### Main Objective
Segment customers based on purchasing behavior using clustering techniques.

---

### Expected Output

You should be able to identify groups such as:

- Loyal Customers  
  → High frequency, high spending  

- Big Spenders  
  → High monetary value  

- New Customers  
  → Recent but low activity  

- Churn Risk Customers  
  → Long time since last purchase  

---

## 7. Learning Goals

By completing this project, you will:

- Understand real-world messy data  
- Perform data cleaning and preprocessing  
- Apply feature engineering (RFM)  
- Handle outliers  
- Perform feature scaling  
- Apply clustering (K-Means)  
- Interpret clusters in business terms  

---

## 8. Key Insight

This dataset focuses on:
> Customer behavior based on transaction history

NOT on:
- Demographics  
- Income  
- Age  

---

## 9. Final Thought

Always ask:
> "What type of customer does each cluster represent?"

This is the most important step in the entire project.
