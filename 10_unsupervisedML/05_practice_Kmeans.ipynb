### k means for iris dataset

iris dataset is dataset of flowers where we have three catrgories of slowers:

1. setosa 2. virginica  3.versicolor

 seloca is perfectly linearly separable where the other two are difficult to separate

```python
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import pandas as pd
```

```python
iris = load_iris()
```

```python
X = iris.data
y = iris.target
```

```python
X.head()
```

```python
#visualize:

sns.scatterplot(x= X[:, 0 ], y=X[:, 2], c=y )
```

```python
#Scaling:

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Elbow method

wcss = []

for k in range(1, 11):
                   kmeans = KMeans(n_clusters=k)
                   kmeans.fit(X_scaled)
                   wcss.append(kmeans.inertia_)
                   
```

```python
sns.lineplot(x=range(1,11), y=wcss, marker='o')
```

```python
# KMEans

kmeans = KMeans(n_clusters = 3)
labels = kmeans.fit_predict(X_scaled)

sns.scatterplot(x=X[:, 0], y=X[:, 2], c=labels)
```
