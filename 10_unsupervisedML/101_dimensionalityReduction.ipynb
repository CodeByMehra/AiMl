# Dimensionality Reduction

Dimensionality Reduction is a technique used in Machine Learning 
to reduce the number of input features (dimensions).

## Problems Due to High Dimensionality

When the number of features increases, several problems occur.
This is known as the **Curse of Dimensionality**.

### 1️⃣ Data Sparsity

- In high-dimensional space, most of the space becomes empty.
- Only a few data points exist.
- Each data point becomes very far from other data points.
- This makes learning difficult .


### 2️⃣ Distance Becomes Meaningless

Many ML algorithms depend on distance between data points:

- Clustering Algorithms
- K-Means
- K-Nearest Neighbors (KNN)

In high dimensions:
- Data points become far from each other.
- Distances become almost similar.
- Model efficiency decreases.


### 3️⃣ Expensive Computation

- ML models take more time to train.
- Computational complexity increases.
- High memory usage.
- Good accuracy may not be achieved.

## Methods to Reduce Dimensions

There are mainly two approaches:

1. Feature Selection
2. Feature Transformation


### 1️⃣ Feature Selection

Select only important features and remove unnecessary ones.

Example:
- Remove redundant features
- Remove highly correlated features

Correlation matrix can be used to identify redundant features.

```python
import pandas as pd
import numpy as np

# Sample dataset
data = {
    'F1': [1, 2, 3, 4, 5],
    'F2': [2, 4, 6, 8, 10],   # Highly correlated with F1
    'F3': [5, 3, 6, 2, 7]
}

df = pd.DataFrame(data)

# Correlation Matrix
corr_matrix = df.corr()
corr_matrix
```

If correlation between two features is very high (e.g., > 0.9),
we can remove one of them.

### 2️⃣ Feature Transformation

Instead of removing features,
we combine or transform features into new ones.

Example:
- Combine F1, F2, F3 into new feature F'
- Principal Component Analysis (PCA)

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Standardize data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# Apply PCA
pca = PCA(n_components=2)
principal_components = pca.fit_transform(scaled_data)

principal_components
```

PCA creates new features (Principal Components)
which are combinations of original features.

These new features:
- Capture maximum variance
- Reduce dimensionality
- Improve model efficiency

### 3️⃣ Naive Bayes (Special Case)

Naive Bayes assumes features are independent.

Because of this assumption:
- It works well even in high dimensions.
- Less affected by curse of dimensionality.
