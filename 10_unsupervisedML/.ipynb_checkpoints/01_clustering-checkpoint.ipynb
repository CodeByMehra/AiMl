## Clustering (Unsupervised Machine Learning)

Clustering is an unsupervised machine learning technique that automatically groups similar and unlabeled data points into clusters based on a similarity metric.

In clustering:

Data does not contain labels

The algorithm tries to discover patterns automatically

Similar data points are grouped into the same cluster

Definition

Clustering is an unsupervised machine learning technique that groups similar unlabeled data points into clusters based on a defined similarity measure.


#  Basic Idea of Clustering

Suppose we have some data points.

For simplicity, we take two features:

Feature  1

Feature 2

Using two features allows us to visualize the data in 2D space, which is easier for humans to understand.

However, this does not mean real datasets only contain two features.

Real-world datasets may have:

3 features

10 features

100 features

even more

These are called multi-dimensional datasets.

# Why We Use 2D Visualization

In many examples, clustering is explained using 2-dimensional plots.

Reason:

Humans can easily understand 2D visualizations

It helps us see how clusters are formed

But in real-world problems:

Data usually contains many dimensions

Clustering algorithms still work on those high-dimensional datasets

So 2D is mainly used for explanation and visualization purposes.
