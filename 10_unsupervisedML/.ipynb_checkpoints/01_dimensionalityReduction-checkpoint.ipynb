# Dimensionality Reduction

Dimensionality Reduction is a technique used in Machine Learning 
to reduce the number of input features (dimensions).

## Problems Due to High Dimensionality

When the number of features increases, several problems occur.
This is known as the **Curse of Dimensionality**.

### 1️⃣ Data Sparsity

- In high-dimensional space, most of the space becomes empty.
- Only a few data points exist.
- Each data point becomes very far from other data points.
- This makes learning difficult.


### 2️⃣ Distance Becomes Meaningless

Many ML algorithms depend on distance between data points:

- Clustering Algorithms
- K-Means
- K-Nearest Neighbors (KNN)

In high dimensions:
- Data points become far from each other.
- Distances become almost similar.
- Model efficiency decreases.


### 3️⃣ Expensive Computation

- ML models take more time to train.
- Computational complexity increases.
- High memory usage.
- Good accuracy may not be achieved.

## Methods to Reduce Dimensions

There are mainly two approaches:

1. Feature Selection
2. Feature Transformation


### 1️⃣ Feature Selection

Select only important features and remove unnecessary ones.

Example:
- Remove redundant features
- Remove highly correlated features

Correlation matrix can be used to identify redundant features.
