# Principal Component Analysis (PCA)

PCA stands for **Principal Component Analysis**.

It is a dimensionality reduction technique that:
- Combines features
- Reduces dimensions
- Keeps maximum information

#  Intuition Example (OTT Platform)

Suppose we are analyzing user engagement on an OTT platform.

We have these features:

- Number of shows watched per week
- Average watch time (minutes)
- % completion rate

These 3 features together represent:

 USER ENGAGEMENT

Instead of keeping 3 features,
can we combine them into 1 strong feature?

YES — PCA does exactly this.

# Step 1️⃣: Spread of Data (Variance)

PCA looks at:

👉 Where is the data spread the most?

Because:
More spread = More information
Less spread = Less information
