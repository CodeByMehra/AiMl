## DBSCAN for anomalies:

- in DBSCAN, the most technical definition for the noise is: For a neighbourhood there should be the number of samples < min samples required

- as the DBSAN is able to make clusters as based on the number of samples, it can easily detect anomalies

```python
# Code 
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```

```python
X, y = make_moons(
    n_samples = 500,
    noise =0.1,
    random_state = 42
)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Visualising the scaled data

sns.scatterplot(x=X_scaled[:, 0], y = X_scaled[:, 1])
```

### to choose value for minimum samples: 

- 2*number of features can be the good value for minimum samples (Featuers can me known by X_xcaled.shape), also the size of total data is considered

### guide to choose the epsilon value: 

- if there is low dimension (<= 10 features) -> eps can be 0.5 - 1

- if there is medium dimension (10 - 20 features) -> eps can be 1 - 2

- if there is high dimension (>= 10 features) -> eps can be > 2

```python
# DBSCAN 
dbscan = DBSCAN(
    eps= 0.5 ,
    min_samples= 5
)
labels = dbscan.fit_predict(X_scaled) 
```

```python
sns.scatterplot(x=X_scaled[:, 0], y = X_scaled[:, 1], c=labels)
```

```python
# as we can see here the sbscan made the whole data as a singel cluster, means we gave a large value of epsilon
```

```python
dbscan = DBSCAN(
    eps= 0.2 ,
    min_samples= 5
)
new_labels = dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:, 0], y = X_scaled[:, 1], c=new_labels)
```
