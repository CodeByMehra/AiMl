# Unsupervised Learning – Anomaly Detection

In this section, we learn about **Anomaly Detection**.

Anomalies are unusual data points in a dataset.
They are often confused with outliers, but they are not exactly the same.

## 🔹 What are Anomalies?

Anomalies = Data points that are significantly different from the majority of data.

In simple words:
An anomaly is something unusual or rare.

Examples:
- Fraud transaction in banking
- Defective product in manufacturing
- Network intrusion in cybersecurity

## 🔹 Outliers vs Anomalies

Before this (in supervised learning), we often removed outliers.

Why?

Because:
Outliers can introduce bias in the model.
They may distort the training process.

But in anomaly detection:
We DON'T remove them.
We DETECT them.

Very important conceptual shift:
Earlier → Remove extreme values
Now → Detect and analyze extreme values


## 🔹 What is an Outlier?

An outlier is a data point that is very different from other data points.

It may represent:
- Extreme values
- Measurement errors
- Rare events



## 🎯 Example: Student Study Hours vs Score

Suppose:
X-axis → Study Hours
Y-axis → Exam Score

Normally:
More study hours → Higher score

Most students may:
Study up to 6 hours
Score up to 80 marks

So majority of points lie in that region.

```python
import numpy as np
import matplotlib.pyplot as plt

# Majority students
np.random.seed(42)
study_hours = np.random.uniform(1, 6, 50)
scores = study_hours * 12 + np.random.normal(0, 5, 50)

# Plot majority data
plt.scatter(study_hours, scores)
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score (Normal Data)")
plt.show()
```

## 🔹 Now Suppose There is an Extreme Data Point

Imagine one student:
Studies 1 hour
But scores 95 marks

This point does not follow the normal pattern.

This is:
- Different from other points
- Does not follow trend
- Rare

This could be an outlier or anomaly.

```python
# Add extreme point
study_hours = np.append(study_hours, 1)
scores = np.append(scores, 95)

plt.scatter(study_hours, scores)
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score (With Outlier)")
plt.show()
```

## 🔥 Important Understanding

In Supervised Learning:
We often remove such points.

Because:
They may negatively affect regression line or classification boundary.

In Unsupervised Learning (Anomaly Detection):
Our goal is to detect such unusual patterns.

We want to ask:
Why is this student scoring 95 with 1 hour study?
Is it:
- Error?
- Cheating?
- Genius?
- Data issue?
