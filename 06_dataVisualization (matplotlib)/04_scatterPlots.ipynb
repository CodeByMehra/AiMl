```python
# Scatter Plots
import matplotlib.pyplot as plt
people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

#seeing correlation of bp according to Age
plt.scatter(age,bp, s = bp, cmap = "plasma", c=bp) #plasma is inbuuild color map and c value is the value by respect of which the color will change

plt.title("age vs bp")
plt.xlabel("Age")
plt.ylabel("Bp")

plt.grid()
```

```python
#adding annotations:

people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

# syntax: plt.annotate("text",xy= (x_point, y_point), xytext = (x_text, y_text))

for i in range(len(people)):
    plt.annotate(people[i], xy= (age[i],bp[i]), xytext= (age[i],bp[i])) 

plt.scatter(age,bp, s = bp, cmap = "plasma", c=bp) #plasma is inbuuild color map and c value is the value by respect of which the color will change

plt.title("age vs bp")
plt.xlabel("Age")
plt.ylabel("Bp")

plt.grid()
```
