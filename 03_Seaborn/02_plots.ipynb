```python
import matplotlib.pyplot as plt
import seaborn as sns
```

# plotting different types of plots in seaborn:


# Relational plots

```python
# Relational plots
# scatter plot
sns.get_dataset_names()
tips = sns.load_dataset("tips")
print(tips.head())

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
plt.show()
```

```python
# line plot
flights = sns.load_dataset("flights")

sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
plt.show()
```

# Categorical plots

```python
# Bar plot
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)
plt.show()
```

```python
# Box plot
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"
)
plt.show()
```
